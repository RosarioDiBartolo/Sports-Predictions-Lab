# 02 · Calibrazione del mercato

Misuriamo accuracy, log loss, Brier score ed Expected Calibration Error. La diagonale della curva rappresenta una calibrazione perfetta.

```python
from pathlib import Path
import sys

PROJECT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_DIR / "src"))

import matplotlib.pyplot as plt
import pandas as pd
from football_odds import AnalysisConfig, run_analysis
```

```python
config = AnalysisConfig(project_dir=PROJECT_DIR, bin_width=0.05)
result = run_analysis(config)
pd.Series(result.metrics, name="valore").to_frame()
```

```python
result.season_metrics.style.format({
    "accuracy": "{:.2%}",
    "log_loss": "{:.4f}",
    "brier_score": "{:.4f}",
    "average_margin": "{:.2%}",
})
```

```python
result.calibration.style.format({
    "predicted_probability": "{:.1%}",
    "actual_frequency": "{:.1%}",
    "calibration_error": "{:+.1%}",
})
```

```python
table = result.calibration
fig, ax = plt.subplots(figsize=(8, 8))
ax.plot([0, 1], [0, 1], "--", color="gray", label="Calibrazione perfetta")
ax.plot(table["predicted_probability"], table["actual_frequency"], "o-", label="Mercato")
ax.set(xlabel="Probabilità prevista", ylabel="Frequenza reale", xlim=(0, 1), ylim=(0, 1))
ax.grid(alpha=0.3)
ax.legend();
```

## Lettura del risultato

- Punti sopra la diagonale: l'evento accade più spesso di quanto indicato dalle quote.
- Punti sotto la diagonale: il mercato lo sovrastima.
- ECE, log loss e Brier più bassi indicano prestazioni migliori.
- Le fasce con poche osservazioni vanno interpretate con cautela.
